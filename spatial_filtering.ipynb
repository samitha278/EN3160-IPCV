```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import cv2
import matplotlib.pyplot as plt
```

## Convolution

```python
torch.manual_seed(278)
image = torch.randn((12,12)).view(1,1,12,12)
kernel = (1/9) * torch.ones((3,3)).view(1,1,3,3)
image,kernel
```

```python
out = F.conv2d(image,kernel)
out
```

```python
conv_layer = nn.Conv2d(1,1,3)
out1 = conv_layer(image)
out1
```
